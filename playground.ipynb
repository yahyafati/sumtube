```python
from youtube_transcript_api import YouTubeTranscriptApi, FetchedTranscript
import re
```

```python
def get_video_id(url):
    """Extract video ID from various YouTube URL formats."""
    pattern = r"(?:v=|\/)([0-9A-Za-z_-]{11}).*"
    match = re.search(pattern, url)
    if not match:
        raise ValueError("Invalid YouTube URL")
    return match.group(1)

def get_youtube_transcript(url: str, languages=None):
    if languages is None:
        languages = ['en']
    video_id = get_video_id(url)
    ytt_api = YouTubeTranscriptApi()
    transcript = ytt_api.fetch(video_id, languages=languages)
    return transcript
    # print(transcript)
    # text = " ".join([t['text'] for t in transcript])
    # return text

def join_transcript_snippets(transcript: FetchedTranscript) -> str:
    return " ".join(snippet.text for snippet in transcript.snippets)
```

```python
import os
import shutil

def copy_markdown_files(src_folder, dst_folder):
    os.makedirs(dst_folder, exist_ok=True)

    for root, _, files in os.walk(src_folder):
        for file in files:
            if file.endswith(".md"):
                src_path = os.path.join(root, file)
                dst_path = os.path.join(dst_folder, file)

                # If duplicate filenames exist, rename by appending a counter
                base, ext = os.path.splitext(file)
                counter = 1
                while os.path.exists(dst_path):
                    dst_path = os.path.join(dst_folder, f"{base}_{counter}{ext}")
                    counter += 1

                shutil.copy2(src_path, dst_path)
                print(f"Copied: {src_path} -> {dst_path}")
```

```python
src = input("Enter source folder: ").strip()
dst = input("Enter destination folder: ").strip()
copy_markdown_files(src, dst)
```

```python
import os

def merge_markdown(folder_path, output_file):
    """
    Recursively collects all .md files under folder_path and merges them
    into output_file. Each file is preceded by an H1 header with its name.
    """
    merged_parts = []

    for root, _, files in os.walk(folder_path):
        for f in files:
            if f.lower().endswith(".md"):
                full_path = os.path.join(root, f)
                with open(full_path, "r", encoding="utf-8") as md:
                    content = md.read()
                content = content.strip()
                # header = f"# {f}\n\n"
                header = ""
                merged_parts.append(header + content + "\n\n")

    # Write final merged file
    with open(output_file, "w", encoding="utf-8") as out:
        out.write("".join(merged_parts))
```

```python
import os
from pathlib import Path

def merge_markdown(folder_path, output_file):
    """
    Recursively collects all .md files under folder_path (alphabetically)
    and merges them into output_file. Each file is preceded by an H1 header.
    """
    markdown_paths = []

    # Collect markdown files with full paths
    for root, dirs, files in os.walk(folder_path):
        dirs.sort()          # ensure deterministic alphabetical directory order
        files = sorted(files, key=str.lower)
        for f in files:
            if f.lower().endswith(".md"):
                markdown_paths.append(os.path.join(root, f))

    merged_parts = []

    # Merge in alphabetical order (full relative path ordering)
    markdown_paths.sort(key=lambda p: os.path.basename(p))

    for path in markdown_paths:
        filename = os.path.basename(path)
        with open(path, "r", encoding="utf-8") as md:
            content = md.read()
        content = content.strip()
        content = f"{content} \n\n---\n\n"
        # content = f"# {filename}\n\n" + content + "\n\n"
        merged_parts.append(content)

    with open(output_file, "w", encoding="utf-8") as out:
        out.write("".join(merged_parts))
```
